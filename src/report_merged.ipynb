```python
# =========================
# 基于实际表头收窄产品报表字段
# 只保留：order_sn（有则保留）、id_card_no_md5/证件号键、report_time、产品输出/命中标识字段
# =========================
# 这些字段是各产品表里的请求/流水/更新时间/姓名手机等元信息，不作为产品输出字段保留。
COMMON_META_COLS = {
    'report_id', 'merchant_code', 'id_card_name_md5', 'mobile_md5', 'swift_number',
    'batch_no', 'created_at', 'updated_at',
    'id_card_name', 'mobile',
    'md5_id_card_name', 'md5_receiver_mobile', 'seqNum',
}

# 我已按实际表头判断：
# - 回溯产品表时间字段为 report_time。
# - 仟盈/仟诺/信用司南/玉明恒时间字段为 dispatch_time，输出中统一改名为 report_time。
# - md5_id_card_no 输出中统一改名为 id_card_no_md5。
# - 布尔天狼星表没有 id_card_no_md5，只有 id_card_no，这里保留原字段，不伪装成 md5。
PRODUCT_REPORT_CONFIG = {
    'bairong_blackgraylevel_risk_report': {
        'df': bairong_blackgraylevel_risk_report, 
        'key_cols': ['order_sn', 'id_card_no_md5'],
        'time_col': 'report_time',
    },
    # 'bairong_fraud_relation_g_report': {
    #     'df': bairong_fraud_relation_g_report,
    #     'key_cols': ['order_sn', 'id_card_no_md5'],
    #     'time_col': 'report_time',
    # },
    'bairong_scorefxsbindex_norm_risk_report': {
        'df': bairong_scorefxsbindex_norm_risk_report,
        'key_cols': ['order_sn', 'id_card_no_md5'],
        'time_col': 'report_time',
    },
    'bairong_scorefxsbindex_qlty_risk_report': {
        'df': bairong_scorefxsbindex_qlty_risk_report,
        'key_cols': ['order_sn', 'id_card_no_md5'],
        'time_col': 'report_time',
    },
    'bairong_score_risk_report': {
        'df': bairong_score_risk_report,
        'key_cols': ['order_sn', 'id_card_no_md5'],
        'time_col': 'report_time',
    },
    # 'kfong_normal_score_risk_report': {
    #     'df': kfong_normal_score_risk_report,
    #     'key_cols': ['order_sn', 'id_card_no_md5'],
    #     'time_col': 'report_time',
    # },
    # 'kfong_premium_score_risk_report': {
    #     'df': kfong_premium_score_risk_report,
    #     'key_cols': ['order_sn', 'id_card_no_md5'],
    #     'time_col': 'report_time',
    # },
    # 'kfong_qianying_score_risk_report': {
    #     'df': kfong_qianying_score_risk_report,
    #     'key_cols': ['order_sn', 'id_card_no_md5'],
    #     'time_col': 'report_time',
    # },
    # 'kfong_selection_score_risk_report': {
    #     'df': kfong_selection_score_risk_report,
    #     'key_cols': ['order_sn', 'id_card_no_md5'],
    #     'time_col': 'report_time',
    # },
    # 'nuoer_score_risk_report': {
    #     'df': nuoer_score_risk_report,
    #     'key_cols': ['order_sn', 'id_card_no_md5'],
    #     'time_col': 'report_time',
    # },
    'qianying_test_result': {
        'df': qianying_test_result,
        'key_cols': ['md5_id_card_no'],
        'time_col': 'dispatch_time',
    },
    'qiannuo_test_result': {
        'df': qiannuo_test_result,
        'key_cols': ['md5_id_card_no'],
        'time_col': 'dispatch_time',
    },
    'yumingheng_test_sample': {
        'df': yumingheng_test_sample,
        'key_cols': ['md5_id_card_no'],
        'time_col': 'dispatch_time',
    },
    'buer_tlx': {
        'df': buer_tlx,
        'key_cols': ['order_sn', 'id_card_no'],
        'time_col': 'report_time',
    },
    'credit_sinan_xingyao': {
        'df': credit_sinan_xingyao,
        'key_cols': ['md5_id_card_no'],
        'time_col': 'dispatch_time',
        'extra_drop_cols': ['返回结果状态码'],  # 如果你希望保留接口返回状态，可以删掉这一行。
    },
}


def _existing_cols(df, cols):
    return [col for col in cols if col in df.columns]


def get_product_output_cols(df, key_cols, time_col, extra_drop_cols=None):
    drop_cols = set(COMMON_META_COLS)
    drop_cols.update(key_cols)
    drop_cols.add(time_col)
    if extra_drop_cols:
        drop_cols.update(extra_drop_cols)
    return [col for col in df.columns if col not in drop_cols]


def slim_product_report_by_schema(table_name, df, key_cols, time_col, extra_drop_cols=None):
    key_cols = _existing_cols(df, key_cols)
    if time_col not in df.columns:
        raise KeyError(f'{table_name} 找不到时间字段: {time_col}')

    product_cols = get_product_output_cols(
        df,
        key_cols=key_cols,
        time_col=time_col,
        extra_drop_cols=extra_drop_cols,
    )
    keep_cols = key_cols + [time_col] + product_cols
    slim_df = df.loc[:, keep_cols].copy()

    rename_cols = {}
    if time_col != 'report_time':
        rename_cols[time_col] = 'report_time'
    if 'md5_id_card_no' in slim_df.columns:
        rename_cols['md5_id_card_no'] = 'id_card_no_md5'
    slim_df = slim_df.rename(columns=rename_cols)

    return slim_df, product_cols


product_slim_tables = {}
product_keep_catalog_rows = []

for table_name, config in PRODUCT_REPORT_CONFIG.items():
    slim_df, product_cols = slim_product_report_by_schema(
        table_name=table_name,
        df=config['df'],
        key_cols=config['key_cols'],
        time_col=config['time_col'],
        extra_drop_cols=config.get('extra_drop_cols'),
    )
    slim_name = f'{table_name}_slim'
    product_slim_tables[slim_name] = slim_df
    globals()[slim_name] = slim_df

    product_keep_catalog_rows.append({
        'table_name': table_name,
        'slim_name': slim_name,
        'raw_cols': config['df'].shape[1],
        'keep_cols': slim_df.shape[1],
        'key_cols': ', '.join([col for col in ['order_sn', 'id_card_no_md5', 'id_card_no'] if col in slim_df.columns]),
        'time_col': 'report_time',
        'product_cols_cnt': len(product_cols),
        'product_cols': ', '.join(product_cols),
    })

product_keep_catalog = pd.DataFrame(product_keep_catalog_rows)
product_keep_catalog[['table_name', 'slim_name', 'raw_cols', 'keep_cols', 'key_cols', 'time_col', 'product_cols_cnt', 'product_cols']]
```

```python
# todo 合并 slim 报告表：先 order_sn，未命中再用 id_card_no_md5 补
REPORT_MAX_DAYS = 3

if 'df' not in globals():
    raise NameError('请先读取用户样本 df')

df_report_merged = df.copy()

base_rename_cols = {
    src: dst
    for src, dst in {
        'root_order_sn': 'order_sn',
        'root_order_time': 'order_time',
        'root_start_date': 'start_date',
        'root_rent_days': 'rent_days_sum',
        'final_label': 'target',
    }.items()
    if dst not in df_report_merged.columns and src in df_report_merged.columns
}
if base_rename_cols:
    df_report_merged = df_report_merged.rename(columns=base_rename_cols)

required_base_cols = {'order_sn', 'id_card_no_md5', 'order_time'}
missing_base_cols = required_base_cols - set(df_report_merged.columns)
if missing_base_cols:
    raise KeyError(f'df 缺少合并必需字段: {sorted(missing_base_cols)}')

df_report_merged['order_sn'] = df_report_merged['order_sn'].astype('string').str.strip()
df_report_merged['id_card_no_md5'] = df_report_merged['id_card_no_md5'].astype('string').str.strip().str.lower()
df_report_merged['order_time'] = pd.to_datetime(df_report_merged['order_time'], errors='coerce')

REPORT_SLIM_TABLES_TO_MERGE = [
    ('bairong_blackgraylevel_risk_report', bairong_blackgraylevel_risk_report_slim),
    ('bairong_scorefxsbindex_norm_risk_report', bairong_scorefxsbindex_norm_risk_report_slim),
    ('bairong_scorefxsbindex_qlty_risk_report', bairong_scorefxsbindex_qlty_risk_report_slim),
    ('bairong_score_risk_report', bairong_score_risk_report_slim),
    ('qianying_test_result', qianying_test_result_slim),
    ('qiannuo_test_result', qiannuo_test_result_slim),
    ('yumingheng_test_sample', yumingheng_test_sample_slim),
    ('buer_tlx', buer_tlx_slim),
    ('credit_sinan_xingyao', credit_sinan_xingyao_slim),
]


def _prepare_report_for_merge(report_df, order_col='order_sn', id_col='id_card_no_md5', time_col='report_time'):
    report = report_df.copy()

    if id_col not in report.columns and 'md5_id_card_no' in report.columns:
        report = report.rename(columns={'md5_id_card_no': id_col})
    if id_col not in report.columns and 'id_card_no' in report.columns:
        report[id_col] = report['id_card_no'].apply(md5_id_card)

    if order_col in report.columns:
        report[order_col] = report[order_col].astype('string').str.strip()
    if id_col in report.columns:
        report[id_col] = report[id_col].astype('string').str.strip().str.lower()
    if time_col in report.columns:
        report[time_col] = pd.to_datetime(report[time_col], errors='coerce')

    return report.drop(columns=['id_card_no'], errors='ignore')


def _rename_conflicting_report_features(df_left, report, table_name, order_col='order_sn', id_col='id_card_no_md5', time_col='report_time'):
    key_cols = {order_col, id_col, time_col}
    rename_cols = {}
    used_cols = set(df_left.columns) | set(report.columns)

    for col in report.columns:
        if col in key_cols or col not in df_left.columns:
            continue

        candidate = f'{table_name}__{col}'
        idx = 2
        while candidate in used_cols or candidate in rename_cols.values():
            candidate = f'{table_name}__{col}_{idx}'
            idx += 1
        rename_cols[col] = candidate
        used_cols.add(candidate)

    return report.rename(columns=rename_cols), rename_cols


def merge_one_slim_report(df_left, report_df, table_name, max_days=REPORT_MAX_DAYS):
    order_col = 'order_sn'
    id_col = 'id_card_no_md5'
    left_time_col = 'order_time'
    right_time_col = 'report_time'
    match_type_col = f'{table_name}_match_type'

    report = _prepare_report_for_merge(report_df, order_col=order_col, id_col=id_col, time_col=right_time_col)
    if right_time_col not in report.columns:
        raise KeyError(f'{table_name} 缺少报告时间字段: {right_time_col}')
    if id_col not in report.columns:
        raise KeyError(f'{table_name} 缺少证件号 md5 字段: {id_col}')

    report, renamed_cols = _rename_conflicting_report_features(
        df_left=df_left,
        report=report,
        table_name=table_name,
        order_col=order_col,
        id_col=id_col,
        time_col=right_time_col,
    )
    feature_cols = [col for col in report.columns if col not in {order_col, id_col, right_time_col}]

    if order_col in report.columns:
        merged, order_key, idcard_key = match_report_order_then_idcard(
            df_user=df_left,
            df_report_order=report,
            order_col=order_col,
            id_col=id_col,
            left_time_col=left_time_col,
            right_time_col=right_time_col,
            max_days=max_days,
            match_type_col=match_type_col,
        )
    else:
        merged, idcard_key = match_nearest_report_by_idcard(
            df_user=df_left,
            df_br_card=report,
            join_col=id_col,
            left_time_col=left_time_col,
            right_time_col=right_time_col,
            max_days=max_days,
        )
        order_key = pd.DataFrame()
        merged[match_type_col] = pd.NA
        if len(idcard_key) > 0:
            idcard_match_flag = (
                idcard_key[[id_col, left_time_col]]
                .drop_duplicates()
                .assign(_idcard_match=True)
            )
            merged = merged.merge(idcard_match_flag, on=[id_col, left_time_col], how='left')
            merged.loc[merged['_idcard_match'].eq(True), match_type_col] = id_col
            merged = merged.drop(columns=['_idcard_match'])

    matched_rows = int(merged[feature_cols].notna().any(axis=1).sum()) if feature_cols else 0
    summary = {
        'table_name': table_name,
        'input_rows': len(report_df),
        'feature_cols': len(feature_cols),
        'order_match_rows': len(order_key),
        'idcard_fill_rows': len(idcard_key),
        'matched_rows': matched_rows,
        'renamed_feature_cols': ', '.join([f'{old}->{new}' for old, new in renamed_cols.items()]),
    }

    merged = merged.drop(
        columns=[right_time_col, 'time_diff', 'time_diff_fill', 'fill_type', match_type_col],
        errors='ignore',
    )
    return merged, summary


report_merge_summary_rows = []
for table_name, report_df in REPORT_SLIM_TABLES_TO_MERGE:
    df_report_merged, summary_row = merge_one_slim_report(
        df_left=df_report_merged,
        report_df=report_df,
        table_name=table_name,
        max_days=REPORT_MAX_DAYS,
    )
    report_merge_summary_rows.append(summary_row)

report_merge_summary = pd.DataFrame(report_merge_summary_rows)
report_merge_summary
```

```python
# =========================
# 1. 读取数据
# =========================
df_oot = pd.read_excel(r'../data/oot用户标签验证.xlsx')
df_qy = pd.read_excel(r'../data/report_file/仟盈分-测试结果.xlsx')
df_qn = pd.read_excel(r'../data/report_file/仟诺分-测试结果.xlsx')
df_br = pd.read_excel(r'../data/report_file/dwd_risk_report_buer_tlx.xlsx')


# =========================
# 2. 保留需要字段
# =========================
df_qy = df_qy[
    ['md5_id_card_no', 'dispatch_time', '仟盈信用分V1', '仟盈通用反欺诈评分V4']
].copy()

df_qn = df_qn[
    ['md5_id_card_no', 'dispatch_time', 'zhili_d4_8', 'beta_pro_v1', 'zc_score01']
].copy()

df_br = df_br[
    ['order_sn', 'id_card_no', 'report_time',
     'apply_time1', 'apply_time12', 'm12_apply_time', 'lend_number12', 'repay_succ6']
].copy()


# =========================
# 3. 参数
# =========================
REPORT_MAX_DAYS = 3   # 和参考文件保持一致：报告时间和订单时间最多相差 3 天


# =========================
# 4. 工具函数
# =========================
def normalize_order_sn(s):
    """
    处理 Excel 读取 order_sn 后可能出现的 .0、空格等问题
    """
    return (
        s.astype('string')
        .str.strip()
        .str.replace(r'\.0$', '', regex=True)
    )


def normalize_md5(s):
    return (
        s.astype('string')
        .str.strip()
        .str.lower()
    )


def prepare_base_df(df_base):
    """
    对 df_oot 做基础字段统一。
    必须最终具备：order_sn、id_card_no_md5、order_time
    """
    df_base = df_base.copy()

    # 兼容 root 字段
    rename_map = {
        'root_order_sn': 'order_sn',
        'root_order_time': 'order_time',
        'root_start_date': 'start_date',
        'root_rent_days': 'rent_days_sum',
        'final_label': 'target',
    }

    real_rename = {
        src: dst
        for src, dst in rename_map.items()
        if src in df_base.columns and dst not in df_base.columns
    }

    if real_rename:
        df_base = df_base.rename(columns=real_rename)

    # 如果只有原始身份证号，没有 md5，则生成 md5
    if 'id_card_no_md5' not in df_base.columns and 'id_card_no' in df_base.columns:
        df_base['id_card_no_md5'] = df_base['id_card_no'].apply(md5_id_card)

    required_cols = {'order_sn', 'id_card_no_md5', 'order_time'}
    missing_cols = required_cols - set(df_base.columns)
    if missing_cols:
        raise KeyError(f'df_oot 缺少合并必需字段: {sorted(missing_cols)}，当前字段为: {df_base.columns.tolist()}')

    df_base['order_sn'] = normalize_order_sn(df_base['order_sn'])
    df_base['id_card_no_md5'] = normalize_md5(df_base['id_card_no_md5'])
    df_base['order_time'] = pd.to_datetime(df_base['order_time'], errors='coerce')

    return df_base


def prepare_report_for_merge(report_df, time_col='report_time'):
    """
    参考 merged_report.py 的处理方式：
    - md5_id_card_no 统一成 id_card_no_md5
    - 如果只有 id_card_no，则用 md5_id_card 转成 id_card_no_md5
    - 时间字段转 datetime
    - 删除原始 id_card_no，避免泄露和重复
    """
    report = report_df.copy()

    if 'id_card_no_md5' not in report.columns and 'md5_id_card_no' in report.columns:
        report = report.rename(columns={'md5_id_card_no': 'id_card_no_md5'})

    if 'id_card_no_md5' not in report.columns and 'id_card_no' in report.columns:
        report['id_card_no_md5'] = report['id_card_no'].apply(md5_id_card)

    if 'order_sn' in report.columns:
        report['order_sn'] = normalize_order_sn(report['order_sn'])

    if 'id_card_no_md5' in report.columns:
        report['id_card_no_md5'] = normalize_md5(report['id_card_no_md5'])

    if time_col in report.columns:
        report[time_col] = pd.to_datetime(report[time_col], errors='coerce')

    report = report.drop(columns=['id_card_no'], errors='ignore')

    return report


def rename_conflicting_report_features(df_left, report, table_name):
    """
    如果报告表字段和主表已有字段重名，就加 table_name 前缀。
    例如：zc_score01 已经存在，则改成 qiannuo_test_result__zc_score01
    """
    key_cols = {'order_sn', 'id_card_no_md5', 'report_time'}
    rename_cols = {}
    used_cols = set(df_left.columns) | set(report.columns)

    for col in report.columns:
        if col in key_cols:
            continue

        if col not in df_left.columns:
            continue

        candidate = f'{table_name}__{col}'
        idx = 2

        while candidate in used_cols or candidate in rename_cols.values():
            candidate = f'{table_name}__{col}_{idx}'
            idx += 1

        rename_cols[col] = candidate
        used_cols.add(candidate)

    report = report.rename(columns=rename_cols)

    return report, rename_cols


def merge_one_report(df_left, report_df, table_name, max_days=3):
    """
    复用参考文件夹方法：
    1. 如果报告表有 order_sn：先 order_sn 匹配，没命中的再 id_card_no_md5 补
    2. 如果报告表没有 order_sn：直接 id_card_no_md5 + 时间窗口匹配最近报告
    """
    df_left = df_left.copy()

    report = prepare_report_for_merge(report_df, time_col='report_time')

    if 'report_time' not in report.columns:
        raise KeyError(f'{table_name} 缺少 report_time 字段')

    if 'id_card_no_md5' not in report.columns:
        raise KeyError(f'{table_name} 缺少 id_card_no_md5 字段')

    report, renamed_cols = rename_conflicting_report_features(
        df_left=df_left,
        report=report,
        table_name=table_name
    )

    feature_cols = [
        c for c in report.columns
        if c not in {'order_sn', 'id_card_no_md5', 'report_time'}
    ]

    match_type_col = f'{table_name}_match_type'

    # 有 order_sn：先订单号，再身份证补
    if 'order_sn' in report.columns:
        merged, order_key, idcard_key = match_report_order_then_idcard(
            df_user=df_left,
            df_report_order=report,
            order_col='order_sn',
            id_col='id_card_no_md5',
            left_time_col='order_time',
            right_time_col='report_time',
            max_days=max_days,
            match_type_col=match_type_col
        )

        order_match_rows = len(order_key)
        idcard_fill_rows = len(idcard_key)

    # 没有 order_sn：直接按身份证 + 时间最近报告匹配
    else:
        merged, idcard_key = match_nearest_report_by_idcard(
            df_user=df_left,
            df_br_card=report,
            join_col='id_card_no_md5',
            left_time_col='order_time',
            right_time_col='report_time',
            max_days=max_days
        )

        order_match_rows = 0
        idcard_fill_rows = len(idcard_key)

        merged[match_type_col] = pd.NA

        if len(idcard_key) > 0:
            idcard_match_flag = (
                idcard_key[['id_card_no_md5', 'order_time']]
                .drop_duplicates()
                .assign(_idcard_match=True)
            )

            merged = merged.merge(
                idcard_match_flag,
                on=['id_card_no_md5', 'order_time'],
                how='left'
            )

            merged.loc[merged['_idcard_match'].eq(True), match_type_col] = 'id_card_no_md5'
            merged = merged.drop(columns=['_idcard_match'])

    matched_rows = int(merged[feature_cols].notna().any(axis=1).sum()) if feature_cols else 0

    summary = {
        'table_name': table_name,
        'input_rows': len(report_df),
        'feature_cols': len(feature_cols),
        'order_match_rows': order_match_rows,
        'idcard_fill_rows': idcard_fill_rows,
        'matched_rows': matched_rows,
        'renamed_feature_cols': ', '.join([f'{old}->{new}' for old, new in renamed_cols.items()])
    }

    # 删除过程字段
    merged = merged.drop(
        columns=[
            'report_time',
            'time_diff',
            'time_diff_fill',
            'fill_type',
            match_type_col
        ],
        errors='ignore'
    )

    return merged, summary


# =========================
# 5. 主表预处理
# =========================
df_report_merged = prepare_base_df(df_oot)

print('df_oot 原始行数:', len(df_report_merged))
print('df_oot 原始订单数:', df_report_merged['order_sn'].nunique())


# =========================
# 6. 报告表字段统一
# =========================

# 仟盈：dispatch_time 改成 report_time
df_qy = df_qy.rename(columns={
    'md5_id_card_no': 'id_card_no_md5',
    'dispatch_time': 'report_time'
})

df_qy['id_card_no_md5'] = normalize_md5(df_qy['id_card_no_md5'])
df_qy['report_time'] = pd.to_datetime(df_qy['report_time'], errors='coerce')


# 仟诺：dispatch_time 改成 report_time
df_qn = df_qn.rename(columns={
    'md5_id_card_no': 'id_card_no_md5',
    'dispatch_time': 'report_time'
})

df_qn['id_card_no_md5'] = normalize_md5(df_qn['id_card_no_md5'])
df_qn['report_time'] = pd.to_datetime(df_qn['report_time'], errors='coerce')


# 百融：原始身份证号转 md5
df_br['id_card_no_md5'] = (
    df_br['id_card_no']
    .apply(md5_id_card)
    .astype('string')
    .str.strip()
    .str.lower()
)

df_br['order_sn'] = normalize_order_sn(df_br['order_sn'])
df_br['report_time'] = pd.to_datetime(df_br['report_time'], errors='coerce')


# =========================
# 7. 逐个报告表合并
# =========================
REPORT_TABLES_TO_MERGE = [
    ('qianying_test_result', df_qy),
    ('qiannuo_test_result', df_qn),
    ('buer_tlx', df_br),
]

summary_rows = []

for table_name, report_df in REPORT_TABLES_TO_MERGE:
    before_rows = len(df_report_merged)

    df_report_merged, summary = merge_one_report(
        df_left=df_report_merged,
        report_df=report_df,
        table_name=table_name,
        max_days=REPORT_MAX_DAYS
    )

    after_rows = len(df_report_merged)

    summary['before_rows'] = before_rows
    summary['after_rows'] = after_rows
    summary['row_diff'] = after_rows - before_rows

    summary_rows.append(summary)

    print(f'已合并 {table_name}: {before_rows} -> {after_rows}, 差异 {after_rows - before_rows}')


# =========================
# 8. 合并结果汇总
# =========================
report_merge_summary = pd.DataFrame(summary_rows)

print('\n合并汇总:')
display(report_merge_summary)

print('\n最终结果:')
print('最终行数:', len(df_report_merged))
print('最终订单数:', df_report_merged['order_sn'].nunique())


# =========================
# 9. 检查是否样本膨胀
# =========================
if len(df_report_merged) != len(df_oot):
    print('\n警告：合并后行数和原始 df_oot 不一致')
    print('df_oot 行数:', len(df_oot))
    print('df_report_merged 行数:', len(df_report_merged))
else:
    print('\n行数检查通过：合并后没有样本膨胀')


# =========================
# 10. 匹配率检查
# =========================
check_cols = [
    '仟盈信用分V1',
    '仟盈通用反欺诈评分V4',
    'zhili_d4_8',
    'beta_pro_v1',
    'zc_score01',
    'apply_time1',
    'apply_time12',
    'm12_apply_time',
    'lend_number12',
    'repay_succ6'
]

check_cols = [c for c in check_cols if c in df_report_merged.columns]

match_info = pd.DataFrame({
    '字段': check_cols,
    '非空数量': [df_report_merged[c].notna().sum() for c in check_cols],
    '匹配率': [df_report_merged[c].notna().mean() for c in check_cols]
})

display(match_info)


# =========================
# 11. 导出结果
# =========================
output_path = r'../data/oot用户标签验证_合并评分报告.xlsx'

with pd.ExcelWriter(output_path, engine='openpyxl') as writer:
    df_report_merged.to_excel(writer, sheet_name='合并结果', index=False)
    report_merge_summary.to_excel(writer, sheet_name='合并汇总', index=False)
    match_info.to_excel(writer, sheet_name='字段匹配率', index=False)

print('\n已导出:', output_path)
```
